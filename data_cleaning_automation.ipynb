```python
import pandas as pd

data = {
    'Customer': ['A', 'B', 'C', 'C', 'D', None],
    'Sales': [1000, 2000, None, None, 4000, 5000],
    'Region': ['North', 'South', 'East', 'East', None, 'West']
}

df = pd.DataFrame(data)

df
```

```python
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()

df
```

```python
df['Sales'] = df['Sales'].fillna(df['Sales'].mean())

df['Region'] = df['Region'].fillna('Unknown')

df['Customer'] = df['Customer'].fillna('Guest')

df
```

```python
print(df.describe())
```

```python
df = df.copy()

df.loc[:, 'Sales'] = df['Sales'].fillna(df['Sales'].mean())

df.loc[:, 'Region'] = df['Region'].fillna('Unknown')

df.loc[:, 'Customer'] = df['Customer'].fillna('Guest')

df
```

```python
import matplotlib.pyplot as plt

df.groupby('Region')['Sales'].sum().plot(kind='bar')

plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')

plt.show()
```
